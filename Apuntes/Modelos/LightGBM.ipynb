```python
### LightGBM

param_grid = {
    "n_estimators": [100,200],
    "learning_rate": [0.1,0.5,1],
    "max_depth": [5,10,15],
    "min_data_in_leaf": [1,10,20,100],
    "class_weight": ["balanced",None]
}

# random_state en el estimador base: el de GridSearchCV solo controla los splits de CV,
# no la aleatoriedad interna de LightGBM (bagging, feature sampling).
lgb_clf = LGBMClassifier(verbose=-1, random_state=42)

lgb_grid = GridSearchCV(lgb_clf,
                        param_grid= param_grid,
                        cv = 5,
                        scoring= "balanced_accuracy",
                        n_jobs=-1)

lgb_grid.fit(X_train, y_train)
```
