# **1.MATLAB**

```python
import matplotlib.pyplot as plt
plt.style.use("tableau-colorblind10")
import numpy as np
```

```python
x = np.linspace(0, 10, 100)

# 1ª fila, 1ª columna
plt.subplot(2, 3, 1)
plt.plot(x, np.sin(x), "r")
plt.title("Seno")

# 1ª fila, 2ª columna
plt.subplot(2, 3, 2)
plt.plot(x, np.cos(x), "g")
plt.title("Coseno")

# 1ª fila, 3ª columna
plt.subplot(2, 3, 3)
plt.plot(x, np.tan(x), "y")
plt.title("Tangente")

# 2ª fila, 1ª columna
plt.subplot(2, 3, 4)
plt.plot(x, np.exp(x/5))
plt.title("Exp")

# 2ª fila, 2ª columna
plt.subplot(2, 3, 5)
plt.plot(x, np.log(x+1))
plt.title("Log")

# 2ª fila, 3ª columna
plt.subplot(2, 3, 6)
plt.plot(x, np.sqrt(x))
plt.title("Raíz")

plt.tight_layout()
plt.show()
```

# **2.AXES**

```python
fig, ax = plt.subplots(3, 1, sharex="col", sharey="row")

# Seno
ax[0].plot(x, np.sin(x), color="red")
ax[0].set_title("Seno")

# Coseno
ax[1].plot(x, np.cos(x), color="green")
ax[1].set_title("Coseno")

# Tangente
ax[2].plot(x, np.tan(x), color="blue")
ax[2].set_title("Tangente")

plt.tight_layout()
plt.show();
```
