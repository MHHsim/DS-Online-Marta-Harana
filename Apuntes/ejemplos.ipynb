```python
import requests
from bs4 import BeautifulSoup

url = "https://realpython.github.io/fake-jobs/"
response = requests.get(url)
sopa = BeautifulSoup(response.text, "lxml")

tarjetas = sopa.find_all("div", class_="card-content")

for t in tarjetas:
    titulo = t.find("h2", class_="title is-5").text.strip()
    empresa = t.find("h3", class_="subtitle is-6 company").text.strip()
    ubicacion = t.find("p", class_="location").text.strip()
    fecha = t.find("p", class_="is-small has-text-grey").text.strip()

    print(titulo, "-", empresa, "-", ubicacion, fecha)
```

```python
import pandas as pd

datos = {
    "titulo": [],
    "empresa": [],
    "ubicacion": [],
    "fecha": []
}

for t in tarjetas:
    titulo = t.find("h2", class_="title is-5").text.strip()
    empresa = t.find("h3", class_="subtitle is-6 company").text.strip()
    ubicacion = t.find("p", class_="location").text.strip()
    fecha = t.find("time").text.strip()

    datos["titulo"].append(titulo)
    datos["empresa"].append(empresa)
    datos["ubicacion"].append(ubicacion)
    datos["fecha"].append(fecha)

df = pd.DataFrame(datos)
df
```

```python
df["ubicacion"].value_counts().sort_values(ascending=False)
```
