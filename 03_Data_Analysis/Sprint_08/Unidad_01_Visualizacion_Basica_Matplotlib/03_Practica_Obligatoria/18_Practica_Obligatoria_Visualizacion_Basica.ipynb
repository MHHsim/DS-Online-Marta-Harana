```python
df_top_poblacion["poblacion_miles"] = df_top_poblacion["population_total"]/1000
df_top_poblacion["poblacion_miles"]
```

```python
# Figura 1

colores = ["blue", "gray", "red", "olive", "brown"]

plt.figure(figsize = (8,7))

bars = plt.bar(df_top_poblacion["city"], 
        df_top_poblacion["poblacion_miles"], 
        color = colores, 
        alpha = 0.5)

plt.bar_label(bars, fmt="%.2f", padding=3)

plt.ylabel("Habitantes (miles)")

plt.title("Ciudades mas pobladas de california")

plt.ylim(0,5000)

plt.show()
```

```python
# Figura 2

fig, ax = plt.subplots()

font_dict = {
    "family": "Arial",
    "size": 14,
    "weight": "bold",
    "color": "steelblue"
}

ax.hist(df["area_total_km2"],
        bins=50,
        alpha=1,
        color="steelblue",
        )

plt.legend(["Hist Area"], loc="upper right")

ax.set_title("Histograma del área total ocupada por las ciudades")
ax.set_xlabel("Superficie (km$^2$)")
ax.set_ylabel("Número de ciudades");
```

```python
outliers = df.nlargest(2, "area_water_km2")
outliers

# punto 1 
x1 = 122.27
y1 = 1345895/1000

# punto 2
x2 = 88
y2 = 3884307/1000
```

```python
# Figura 3

poblacion_miles = df["population_total"]/1000

area_agua = df["area_water_km2"]

plt.figure(figsize= (8,7)) # tamaño

plt.scatter(area_agua,
            poblacion_miles,
            color="green",
            alpha=0.5,
            s=40) # tamaño de los puntos

plt.xlabel("Área de agua en la ciudad (km$^2$)")
plt.ylabel("Población miles de habitantes")

plt.title("Población por kilómetro cuadrado")

props_flecha = {
    "arrowstyle": "->",
    "connectionstyle": "arc3"
}
plt.annotate("Outliers", xy = (x1, y1), xytext = (50, 2800),
    arrowprops = props_flecha)

plt.annotate("", xy = (x2, y2), xytext = (60, 3000),
    arrowprops = props_flecha)

plt.tight_layout()
plt.show()
```

```python
df
```

```python
# Figura 4

plt.figure(figsize=(8,7))

plt.boxplot(
    df["elevation_ft"],     # Datos que se representan en el diagrama de cajas
    vert=False,             # False = boxplot horizontal (por defecto es vertical)
    patch_artist=True,      # IMPORTANTE: permite colorear la caja (sin esto, no cambia el color)

    boxprops=dict(
        facecolor="red",    # Color de relleno de la caja (IQR)
        color="black"       # Color del borde de la caja
    ),

    medianprops=dict(
        color="black",      # Color de la línea de la mediana
        linewidth=2         # Grosor de la línea de la mediana
    ),

    whiskerprops=dict(
        color="black"       # Color de los "bigotes" (líneas que salen de la caja)
    ),

    capprops=dict(
        color="black"       # Color de las "tapas" al final de los bigotes
    ),

    flierprops=dict(
        marker="o",         # Forma de los outliers (círculo)
        markersize=6,       # Tamaño de los puntos outliers
        markerfacecolor="black",  # Color de relleno de los outliers
        alpha=0.5           # Transparencia de los outliers
    )
)

plt.title("Diagrama de caja de la altitud de las ciudades de California")
plt.xlabel("Altitud (pies)")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# ============================================================
# FIGURA 1 — (0,0)
# ============================================================
ax = axes[0, 0]

colores = ["blue", "gray", "red", "olive", "brown"]

bars = ax.bar(df_top_poblacion["city"], 
              df_top_poblacion["poblacion_miles"], 
              color=colores, 
              alpha=0.5)

ax.bar_label(bars, fmt="%.2f", padding=3)
ax.set_ylabel("Habitantes (miles)")
ax.set_title("Ciudades más pobladas de California")
ax.set_ylim(0, 5000)


# ============================================================
# FIGURA 2 — (0,1)
# ============================================================
ax = axes[0, 1]

font_dict = {
    "family": "Arial",
    "size": 14,
    "weight": "bold",
    "color": "steelblue"
}

ax.hist(df["area_total_km2"],
        bins=50,
        alpha=1,
        color="steelblue")

ax.legend(["Hist Area"], loc="upper right")
ax.set_title("Histograma del área total ocupada por las ciudades")
ax.set_xlabel("Superficie (km$^2$)")
ax.set_ylabel("Número de ciudades")


# ============================================================
# FIGURA 3 — (1,0)
# ============================================================
ax = axes[1, 0]

poblacion_miles = df["population_total"] / 1000
area_agua = df["area_water_km2"]

ax.scatter(area_agua,
           poblacion_miles,
           color="green",
           alpha=0.5,
           s=40)

ax.set_xlabel("Área de agua en la ciudad (km$^2$)")
ax.set_ylabel("Población miles de habitantes")
ax.set_title("Población por kilómetro cuadrado")

props_flecha = {
    "arrowstyle": "->",
    "connectionstyle": "arc3"
}

ax.annotate("Outliers", xy=(x1, y1), xytext=(50, 2800),
            arrowprops=props_flecha)

ax.annotate("", xy=(x2, y2), xytext=(60, 3000),
            arrowprops=props_flecha)


# ============================================================
# FIGURA 4 — (1,1)
# ============================================================
ax = axes[1, 1]

ax.boxplot(
    df["elevation_ft"],     # Datos que se representan en el diagrama de cajas
    vert=False,             # False = boxplot horizontal (por defecto es vertical)
    patch_artist=True,      # IMPORTANTE: permite colorear la caja (sin esto, no cambia el color)

    boxprops=dict(
        facecolor="red",    # Color de relleno de la caja (IQR)
        color="black"       # Color del borde de la caja
    ),

    medianprops=dict(
        color="black",      # Color de la línea de la mediana
        linewidth=2         # Grosor de la línea de la mediana
    ),

    whiskerprops=dict(
        color="black"       # Color de los "bigotes" (líneas que salen de la caja)
    ),

    capprops=dict(
        color="black"       # Color de las "tapas" al final de los bigotes
    ),

    flierprops=dict(
        marker="o",         # Forma de los outliers (círculo)
        markersize=6,       # Tamaño de los puntos outliers
        markerfacecolor="black",  # Color de relleno de los outliers
        alpha=0.5           # Transparencia de los outliers
    )
)

ax.set_title("Diagrama de caja de la altitud de las ciudades de California")
ax.set_xlabel("Altitud (pies)")


# ============================================================
plt.tight_layout()
plt.show()
```
