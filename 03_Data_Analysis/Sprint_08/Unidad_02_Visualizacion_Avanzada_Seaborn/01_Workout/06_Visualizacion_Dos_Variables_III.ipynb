# Visualización Aplicada: Comparando dos variables numéricas

Nos enfocamos ahora ya en visualizar la relación entre dos variables de carácter numérico. Empezamos, como siempre, con nuestras importaciones:

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

df_titanic = pd.read_csv("./data/titanic.csv")
df_vuelos_jun = pd.read_csv("./data/dataset_viajes_jun23.csv")
df_bitcoin = pd.read_csv("./data/BTC.csv")
```

### Gráficos y consideraciones

Nos vamos a centrar en:
- Comparación de diagramas de caja
- Comparación de histogramas 
- Diagramas de dispersión

* **Cuándo**: 
    1. En análisis bivariantes todos son útiles aunque si buscamos relaciones los diagramas de dispersión son interesantes.
    2. En presentaciones: Cuando se están comparando valores de una misma variable en dos tiempos o dos situaciones diferentes, usar la comparación de distribuciones y dispersión (y siempre como un elemento más gráfico que numérico, no abrumar... enfocar). En el caso de dos variables que no miden lo mismo, generalmente buscaremos hablar de la correlación así que usar los diagramas de dispersión marcando el mensaje (pendientes de la correlación, existencia de la correlación incluyendo funciones ideales que la representen, etc). 
  
* **Cuándo no**: Rige lo mismo que para la anterior clasificación, en presentaciones, en general, NO USAR HISTOGRAMAS (salvo que el público lo demande) son muchas barras :-), NI DIAGRAMAS DE CAJAS, si tienes que poner una "caja": usa Violines. Si quieres comparar distribuciones, compara KDE/Densidades, y si lo haces con histrogramas no uses ni muchas ni pocas barras y quita toda referencia numérica salvo la necesaria que apoye el mensaje que se quiera dar.


### Comparación de diagramas de caja


Tiene sentido cuando comparamos variables similares para valores de variables categóricas (como hicimos con "aircompany" e "ingresos" y estábamos comparando ingresos con ingresos), pero también comparando dos fotos en distintos momentos temporales:

```python
df_vuelos_jul = pd.read_csv("./data/dataset_viajes_jul23.csv")
df_vuelos_jul
```

```python
#COMPARACIÓN DE DIAGRAMA DE CAJAS
fig,axs = plt.subplots(2,1,figsize=(10,8))
sns.boxplot(x = df_vuelos_jul["ingresos"], ax = axs[0], color = "red")
axs[0].set_title("Julio")
sns.boxplot(x = df_vuelos_jun["ingresos"], ax = axs[1], color = "green")
axs[1].set_title("Junio")
plt.tight_layout()
```

```python
### COMPARACIÓN DE DIAGRAMAS DE CAJAS ENTRE JULIO Y JUNIO

# 1. Crear figura con 2 filas y 1 columna de subgráficos
fig, axs = plt.subplots(2, 1, figsize=(10, 8))
# axs[0] será el gráfico de julio
# axs[1] será el gráfico de junio

# 2. Boxplot de ingresos de JULIO
sns.boxplot(
    x=df_vuelos_jul["ingresos"],   # Variable numérica a representar
    ax=axs[0],                     # Subgráfico superior
    color="red"                    # Color del boxplot
)
axs[0].set_title("Julio", fontsize=14, weight="bold")   # Título del primer gráfico

# 3. Boxplot de ingresos de JUNIO
sns.boxplot(
    x=df_vuelos_jun["ingresos"],   # Variable numérica a representar
    ax=axs[1],                     # Subgráfico inferior
    color="green"                  # Color del boxplot
)
axs[1].set_title("Junio", fontsize=14, weight="bold")   # Título del segundo gráfico

# 4. Ajustar márgenes para que no se solapen elementos
plt.tight_layout()

# 5. Mostrar gráfico final
plt.show()
```


### Comparación de histrogramas


De igual manera que comparamos las dispersiones por los diagramas de cajas, lo hacemos con los histogramas

```python
#Comparación de histogramas

plt.figure(figsize=(10,5)) # creamos una figura y sobre la figura vamos añadiendo gráficas
sns.histplot(x = "ingresos", data = df_vuelos_jul, kde = True, label = "Julio")
sns.histplot(x = "ingresos", data = df_vuelos_jun, kde = True, label = "Junio")
plt.legend();
```

```python
### COMPARACIÓN DE HISTOGRAMAS ENTRE JULIO Y JUNIO

plt.figure(figsize=(10,5))  
# Creamos una figura horizontal donde colocaremos ambos histogramas superpuestos

# Histograma de JULIO
sns.histplot(
    x="ingresos",
    data=df_vuelos_jul,
    kde=True,              # Línea de densidad para ver la forma de la distribución
    label="Julio",         # Nombre para la leyenda
    color="red",           # Color para diferenciarlo
    alpha=0.5              # Transparencia para que no tape al otro
)

# Histograma de JUNIO
sns.histplot(
    x="ingresos",
    data=df_vuelos_jun,
    kde=True,              # Línea de densidad
    label="Junio",
    color="green",
    alpha=0.5
)

plt.title("Comparación de distribuciones de ingresos: Julio vs Junio",
          fontsize=14, weight="bold")   # Título del gráfico

plt.xlabel("Ingresos (€)", fontsize=12)  # Etiqueta del eje X
plt.ylabel("Frecuencia", fontsize=12)    # Etiqueta del eje Y

plt.legend()   # Mostrar la leyenda para identificar cada mes
plt.tight_layout()
plt.show()
```


### Diagramas de dispersión

Para terminar, recuperamos la comparación de variables o columnas que están en el mismo dataframe y hacen referencia a la misma fila, es decir son características de una misma entidad:

```python
# Diagrama de dispersión para dos variables numéricas

variable_numerica_1 = "age"
variable_numerica_2 = "fare"

plt.figure(figsize=(10,5))
sns.scatterplot(x= variable_numerica_1,
                y= variable_numerica_2,
                data = df_titanic,
               s=50); # tamaño de los puntos
```

```python
### DIAGRAMA DE DISPERSIÓN PARA DOS VARIABLES NUMÉRICAS

variable_numerica_1 = "age"     # Variable que irá en el eje X
variable_numerica_2 = "fare"    # Variable que irá en el eje Y

plt.figure(figsize=(10,5))      # Crear figura horizontal para mejor visualización

sns.scatterplot(
    x=variable_numerica_1,      # Eje X: edad del pasajero
    y=variable_numerica_2,      # Eje Y: tarifa pagada
    data=df_titanic,            # DataFrame con los datos
    s=50,                       # Tamaño de los puntos
    alpha=0.6                   # Transparencia para evitar saturación visual
)

plt.title("Relación entre edad y tarifa pagada", fontsize=14, weight="bold")
plt.xlabel("Edad")
plt.ylabel("Tarifa (€)")

plt.tight_layout()
plt.show()
```

Para terminar vamos a ver un complemento interesante, aunque más para el análisis que para las visualizaciones desde mi punto de vista, que es el joinplot:

```python
#Diagrama Joint para dos variables numéricas

sns.jointplot (x = df_vuelos_jul["ingresos"],
              y = df_vuelos_jul["consumo_kg"],
              color = "#f4D391",
              height = 5);
```

```python
#Diagrama Joint para dos variables numéricas

sns.jointplot (x = df_titanic["fare"],
              y = df_titanic["age"],
              color = "#f4D391",
              height = 5);
```

```python
### DIAGRAMA JOINT PARA DOS VARIABLES NUMÉRICAS

# JointPlot combina:
# - un scatterplot central
# - histogramas marginales en los ejes X e Y

sns.jointplot(
    x=df_vuelos_jul["ingresos"],     # Variable numérica en el eje X
    y=df_vuelos_jul["consumo_kg"],   # Variable numérica en el eje Y
    color="#f4D391",                 # Color suave para el scatter y los histogramas
    height=5                         # Tamaño total del gráfico
)

# Mostrar el gráfico
plt.show()
```
