## 1) Histograma

**Análisis**
- Para ver forma de la distribución (asimetría, colas, picos).
- Para detectar outliers.
- Para comparar la dispersión entre variables.

**Presentación**
- Cuando quieras mostrar cómo se distribuye una variable.
- Evitar poner demasiados ejes o detalles; usarlo solo si aporta al mensaje.

```python
fig, ax = plt.subplots(figsize=(4.5,3))

sns.histplot(
    df_viajes["ingresos"] / 1000,  # variable numérica (aquí en miles)
    bins=100,                      # número de barras
    kde=False,                     # sin curva KDE
    color='r',                     # color del histograma
    ax=ax
)

ax.set_xlabel("Ingresos (millones)")  # etiqueta eje X
ax.set_ylabel("Num_Vuelos")           # etiqueta eje Y
```

## 3) Histograma + KDE (combinado)

**Análisis**
- Es el más completo: barras + curva suave.
- Ideal para exploración inicial.

**Presentación**
- Solo si quieres mostrar forma + volumen de datos.
- Simplificar colores y ejes.

```python
fig, ax = plt.subplots(figsize=(4.5,3))

sns.histplot(
    df_viajes["ingresos"] / 1000,
    bins=100,
    kde=True,                      # activa la curva KDE
    color='r',
    ax=ax
)

ax.set_xlabel("Ingresos (millones)")
ax.set_ylabel("Num_Vuelos")
```

## 4) Boxplot

**Análisis**
- Para detectar outliers rápidamente.
- Para ver mediana, cuartiles, rango intercuartílico (IQR).

**Presentación**
- Solo en contextos técnicos o científicos.
- No recomendable para público general.

```python
sns.boxplot(
    x=df_viajes["ingresos"] / 1000,  # variable numérica
    color="skyblue"                  # color del boxplot
)
```

## 5) Violin plot

**Análisis**
- Para ver distribución + densidad en un solo gráfico.
- Útil si hay multimodalidad.

**Presentación**
- Igual que el boxplot: solo si el público lo entiende.

```python
sns.violinplot(
    x=df_viajes["ingresos"] / 1000,
    color="lightgreen"
)
```

```python
### FRECUENCIAS RELATIVAS ENTRE VARIABLE CATEGÓRICA Y BINARIA

### UN SOLO GRÁFICO DE FRECUENCIAS RELATIVAS CON EL (%)

variable_categorica = "who"
variable_binaria    = "alive"
df = df_titanic

# Calcular tabla de porcentajes
tabla = (
    df.groupby(variable_categorica)[variable_binaria]
      .value_counts(normalize=True)
      .rename("porcentaje")
      .mul(100)
      .reset_index()
)

plt.figure(figsize=(7,5))

ax = sns.barplot(
    data=tabla,
    x=variable_categorica,
    y="porcentaje",
    hue=variable_binaria,
    palette=["orange", "steelblue"],
    edgecolor="black",
    linewidth=0.8
)

plt.title(f"Frecuencias relativas: {variable_categorica} vs {variable_binaria}",
          fontsize=14, weight="bold")
plt.ylabel("%")
plt.xlabel(variable_categorica.capitalize())

# --- Añadir números encima de cada barra ---
for p in ax.patches:
    valor = p.get_height()
    ax.annotate(f"{valor:.1f}%",
                (p.get_x() + p.get_width() / 2, valor),
                ha='center', va='bottom',
                fontsize=10)

plt.tight_layout()
plt.show()
```
